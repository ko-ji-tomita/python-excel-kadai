```python
import pandas as pd
from openpyxl import load_workbook
from openpyxl.styles import PatternFill

#Excelファイルを読み込みデータフレーム作成
df1 = pd.read_excel("2022_年間売上表.xlsx")
df2 = pd.read_excel("2023_年間売上表.xlsx")

#データを連結
df_combined = pd.concat([df1, df2])

#商品と売上年ごとにグループ化し、金額の合計を計算
df_grouped = df_combined.groupby(['商品', '売上年']).agg({'金額（千円）': 'sum'}).reset_index()

#集計結果を新しいExcelファイルに書き込む
output_file = "売上集計表.xlsx"
df_grouped.to_excel(output_file, index=False)

#ヘッダー部分(1行目)の背景色を薄いグレーに設定
book = load_workbook(output_file)
sheet = book.active

#ヘッダー部分(1行目)のセル範囲を選択
header_fill = PatternFill(start_color='F2F2F2', end_color='F2F2F2', fill_type='solid')
for cell in sheet[1]:
    cell.fill = header_fill

#保存
book.save(output_file)
book.close()
```
